```python
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris
data = load_iris()
X = data.data  # Data fitur
y = data.target  # Label kelas

# Buat DataFrame dari fitur dan label
df = pd.DataFrame(X, columns=data.feature_names)  # Menambahkan nama kolom
df['target'] = y  # Menambahkan kolom target
df['target_name'] = df['target'].map({i: name for i, name in enumerate(data.target_names)})  # Tambahkan label kelas

# Simpan ke file Excel
df.to_excel("iris_dataset.xlsx", index=False)

print("Dataset Iris berhasil diekspor ke iris_dataset.xlsx")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris

# Load dataset Iris
data = load_iris()
X = data.data  # Data fitur (4 dimensi)
y = data.target  # Label kelas

print ("Data fitur (X):")
print (X)
print ("Label kelas (y):")
print (y)

# Lakukan PCA untuk mereduksi ke 2 dimensi
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Plot hasil reduksi dimensi
plt.figure(figsize=(8, 6))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', edgecolor='k', alpha=0.7)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Reduksi Dimensi dari 4D ke 2D pada Dataset Iris")
# plt.legend(handles=scatter.legend_elements()[0], labels=iris.target_names)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.decomposition import PCA

# 1. Load dataset Excel
df = pd.read_excel("data_imu.xlsx", header=0)

# 2. Pilih semua kolom feature (semua kolom)
X = df.values

# 3. Cek kolom feature dgn print
print ("Data fitur (X):")
print (X)

# 4. Lakukan PCA untuk mereduksi ke 2 dimensi
pca = PCA(n_components=3)
X_pca = pca.fit_transform(X)

# 5. Plot hasil reduksi dimensi
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d') # This makes it 3D!

# 6.Tampilkan plot hasil reduksi
ax.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c='pink', edgecolors='blue')

ax.set_xlabel('Component 1')
ax.set_ylabel('Component 2')
ax.set_zlabel('Component 3')
ax.set_title('PCA: Reduksi Dimensi dari 6D ke 3D pada Dataset data_imu.xlsx')

plt.show()
```
